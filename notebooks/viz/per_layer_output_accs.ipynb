```python
import pandas as pd
# import statsmodels.formula.api as smf
import seaborn as sns

sns.set_style("whitegrid")
sns.set_theme(rc={
    "axes.facecolor": "white",     # axes (plot area)
    "figure.facecolor": "white",   # figure (canvas)
    "grid.color": "0.9",            # light gray grid
    "figure.figsize": (3, 3)
})

blue = "#1f77b4"
red = "#d62728"
```

```python
# sub
df = pd.read_csv('/home/xstefan3/tmp/pycharm_project_437/notebooks/logs/model_vs_probes_preds_llama3b_sub_210925.csv')
# Mark which rows had correct model predictions
df['Correct result'] = df['model_predictions'] == df['labels']

# Count number of probe layers
num_layers = len([c for c in df.columns if "probe" in c])

# For each layer, add a boolean column for probe predicting the correct output
for col_i in range(1, num_layers):
    df[f"probe_l{col_i}_predicts_output"] = df[f"probe_l{col_i}"] == df["labels"]

# Function to compute accuracy per layer for a subset of df
def per_layer_acc(subdf):
    return (
        subdf[[f"probe_l{i}_predicts_output" for i in range(1, num_layers)]]
        .rename(columns={f"probe_l{i}_predicts_output": i for i in range(1, num_layers)})
        .sum(0) / len(subdf)
    )

# Compute accuracies for correct and incorrect subsets
acc_correct = per_layer_acc(df[df["Correct result"]])
acc_incorrect = per_layer_acc(df[~df["Correct result"]])

fig, ax = plt.subplots()
ax.plot(acc_correct.index, acc_correct.values, color=blue, marker="o", markersize=4, label="Correct")
ax.plot(acc_incorrect.index, acc_incorrect.values, color=red, marker="o", markersize=4, label="Incorrect")

ax.set_ylabel("Accuracy")
ax.set_xlabel("Layer")
ax.set_ylim(-0.05, 1.05)
ax.grid(True, linestyle="--", alpha=0.6)

# Legend with title
# ax.legend(title="Prediction", loc="upper left")

plt.tight_layout()
ax.figure.savefig("per_layer_acc_sub.pdf", format="pdf", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# llama 3b-sub
df = pd.read_csv('/home/xstefan3/tmp/pycharm_project_437/notebooks/logs/model_vs_probes_preds_llama3b_sub_210925.csv')
cols = [col for col in df.columns if 'probe' in col]
for col in cols:
    df[col] = abs(df[col] - df['labels'])

df['Correct result'] =  df['model_predictions'] == df['labels']
df[['Correct result'] + cols]

ax = sns.lineplot(
    data=(
        df[['Correct result'] + cols]
        .groupby('Correct result')
        .mean()
        .rename(columns={col: int(col[len('probe_l'):]) for col in cols})
        .reset_index()
        .melt(id_vars=['Correct result'], var_name='layer', value_name='Absolute error')
    ),
    x='layer',
    y='Absolute error',
    hue='Correct result',
    hue_order=[True, False],
    palette={True: "#1f77b4", False: "#d62728"},
    legend=False
)
ax.figure.savefig("Abs_error_sub_Llama3b.pdf", format="pdf", dpi=300, bbox_inches="tight")
```

```python
df = pd.read_csv('/home/xstefan3/tmp/pycharm_project_437/notebooks/logs/model_vs_probes_preds_llama3b_multi_210925.csv')
df['Correct result'] =  df['model_predictions'] == df['labels']

df_breaks = df.copy(deep=True)
num_layers = len([c for c in df.columns if "probe" in c])

df_breaks["probe_l0"] = -1
for col_i in reversed(range(1, num_layers)):
    df["probe_l%s_breaks_result" % col_i] = (df["probe_l%s" % (col_i-1)] == df["Correct result"]) & (df["probe_l%s" % col_i] != df["Correct result"])
    df["probe_l%s_corrects_result" % col_i] = (df["probe_l%s" % (col_i-1)] != df["Correct result"]) & (df["probe_l%s" % col_i] == df["Correct result"])

ax = (df[["probe_l%s_breaks_result" % i for i in range(1, num_layers)]]
    .rename(columns={"probe_l%s_breaks_result" % i: i for i in range(num_layers)}).sum(0) / len(df)).plot()
ax.set_ylabel("Ratio of all")
ax.set_xlabel("Layer")
ax.figure.savefig("per_layer_error_llama3b_multi.pdf", format="pdf", dpi=300, bbox_inches="tight")
```

```python
pd.Series([(df["probe_l%s_corrects_result" % i].sum() - df["probe_l%s_breaks_result" % i].sum()) for i in range(1, num_layers)]).plot()
```
