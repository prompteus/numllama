```python
import pandas as pd

# math->natural (row means ×100, rounded)
math_to_natural = [100, 99, 98, 93, 92, 84, 84, 77, 67, 73, 68, 71, 56, 55, 57, 45]

# natural->natural (white screenshot, "sin accs")
natural_to_natural = [80, 99, 98, 99, 99, 100, 99, 99, 99, 99, 99, 99, 99, 99, 98, 41]

# natural->math (black screenshot, "sin accs")
natural_to_math = [85, 93, 100, 99, 100, 100, 100, 97, 97, 100, 100, 100, 100, 98, 97, 45]

# math->math (new values, 17 total, we take first 16 for alignment)
math_to_math = [89, 96, 100, 99, 100, 100, 100, 99, 98, 99, 100, 99, 99, 99, 96, 99]

# Construct dataframe
df = pd.DataFrame({
    "math->natural": math_to_natural,
    "natural->natural": natural_to_natural,
    "natural->math": natural_to_math,
    "math->math": math_to_math,
})
df
```

```python
import matplotlib.pyplot as plt

# X-axis is layer index
x = range(len(df))

fig, ax = plt.subplots(figsize=(5.5, 3))

# Plot according to train/eval settings
ax.plot(x, df["natural->natural"], color="orange", linestyle="-", label="Nat. lang → Nat. lang")
ax.plot(x, df["natural->math"], color="orange", linestyle="--", label="Nat. lang → Math")
ax.plot(x, df["math->natural"], color="blue", linestyle="-", label="Math → Nat. lang")
ax.plot(x, df["math->math"], color="blue", linestyle="--", label="Math → Math")

ax.set_xlabel("Layer")
ax.set_ylabel("Accuracy (%)")
# ax.set_title("Probing performance")
ax.grid(True, linestyle="--", alpha=0.6)
ax.legend(ncol=2, loc="lower left")

plt.tight_layout()
plt.show()
```
