```python
data = {
    "TunedLens: numbers": [0.998, 0.498, 0.023, 0.007, 0.005, 0.003],
    "TunedLens: other tokens": [0.995, 0.542, 0.317, 0.209, 0.155, 0.128],
    "Sin probe: numbers": [1.0, 1.0, 0.91, 0.67, 0.05, 0.002],
}
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(data)

soft_colors = ["#006400", "#00008B", "#B8860B"]  # soft blue, red, green
ax = df.plot(figsize=(5, 3.2), grid=True, color=soft_colors, marker="X", markersize=8)
ax.set_xlabel("Recovered token offset")
ax.set_ylabel("Accuracy")
# ax.legend(loc="lower left")

ax.set_xticks(range(len(df)))
ax.set_xticklabels([0] + [-i for i in range(1, len(df))])
ax.invert_xaxis()

plt.tight_layout()
plt.savefig("prev_token_accuracy.pdf", format="pdf")
plt.show()
```
